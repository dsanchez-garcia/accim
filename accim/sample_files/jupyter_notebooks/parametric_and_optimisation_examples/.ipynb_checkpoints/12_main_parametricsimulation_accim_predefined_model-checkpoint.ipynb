# 12 - Main use: ParametricSimulation (`accim predefined model`)

Objetivo: flujo parametrico simple con `ParametricSimulation` para
`parameters_type="accim predefined model"`, usando los mismos IDFs y EPWs
del notebook `01_parametric_accim_custom_model.ipynb`.

Incluye combinaciones discretas (`sampling_full_set`) y visualizacion.


```python
from pathlib import Path

import pandas as pd
import accim
from besos import eppy_funcs as ef
from accim.parametric_and_optimisation.main import ParametricSimulation

pd.set_option("display.max_columns", 160)
```

## Step 1 - Rutas y archivos de entrada


```python
PROJECT_ROOT = Path.cwd()

idf_list = [
    "OSM_TestResidentialUnit_v01_onlygeometry_SchNatVent_v2520_high_performance.idf",
    "OSM_TestResidentialUnit_v01_onlygeometry_SchNatVent_v2520_low_performance.idf",
]
epw_list = ["Seville.epw", "Sydney.epw"]
OUTPUT_FREQ = "Hourly"
RESULTS_DIRNAME = "nb12_main_parametric_predefined_results"

print("IDF files found:", [(PROJECT_ROOT / idf).exists() for idf in idf_list])
print("EPW files found:", [(PROJECT_ROOT / epw).exists() for epw in epw_list])
```

## Step 2 - Cargar modelo, outputs y parametros discretos


```python
building = ef.get_building(idf_list[0])
accim.utils.reduce_runtime(idf_object=building, timesteps=2)

parametric = ParametricSimulation(
    building=building,
    parameters_type="accim predefined model",
    epws=epw_list,
    output_freqs=[OUTPUT_FREQ],
)

df_meters_sel = pd.DataFrame(
    {
        "key_name": ["Heating:Electricity", "Cooling:Electricity"],
        "frequency": [OUTPUT_FREQ, OUTPUT_FREQ],
    }
)

parametric.apply_outputs_preflight(
    df_meters_sel=df_meters_sel,
    df_vars_sel=None,
    clean_mode="all",
    validate_before_apply=False,
    validate_after_apply=True,
    on_missing="raise",
)
parametric.set_outputs_for_simulation(df_output_meter=df_meters_sel)

predef_parameters = {
    "ComfStand": [1, 2],
    "HVACmode": [2],
    "CAT": [1, 2],
    "CATcoolOffset": [0],
    "CATheatOffset": [0],
}
parametric.set_parameters(accis_params_dict=predef_parameters)
parametric.set_problem()
parametric.sampling_full_set()

plan_df = parametric.parameters_values_df.copy()
plan_df["idf"] = [idf_list[i % len(idf_list)] for i in range(len(plan_df))]

print("Plan rows:", len(plan_df))
print(plan_df.head())
```

## Step 3 - Ejecutar simulacion parametrica


```python
RUN_PARAMETRIC_SIMULATION = True

if RUN_PARAMETRIC_SIMULATION:
    outputs = parametric.run_parametric_simulation(
        out_dir=RESULTS_DIRNAME,
        df=plan_df,
        processes=1,
        keep_input=True,
        keep_dirs=False,
    )
    print("Rows:", len(outputs))
    print(outputs.head())
else:
    outputs = pd.DataFrame()
    print("Simulation skipped. Set RUN_PARAMETRIC_SIMULATION = True to execute the run.")
```

## Step 4 - Normalizacion (AnalysisMixin) y visualizacion (PlottingMixin)


```python
if not outputs.empty:
    parametric.set_building_floor_area(mode="all")
    parametric.normalize_outputs(df_types=["parametric"])
    norm_df = parametric.outputs_param_simulation.copy()

    preview_cols = [c for c in ["idf", "epw", "ComfStand", "CAT"] if c in norm_df.columns]
    energy_cols = [c for c in norm_df.columns if "Heating:Electricity" in c or "Cooling:Electricity" in c]
    print(norm_df[preview_cols + energy_cols[:2]].head())

    plot_out_dir = f"{RESULTS_DIRNAME}/plots"
    plot_col = "ComfStand" if "ComfStand" in norm_df.columns else None
    plot_row = "CAT" if "CAT" in norm_df.columns else None
    plot_hue = None

    parametric.plot_categorical_boxplots(
        df_source="parametric",
        col=plot_col,
        row=plot_row,
        hue=plot_hue,
        out_dir=plot_out_dir,
        normalize_per_m2=False,
        show_points=True,
        sharey=True,
        height=3.6,
        aspect=1.1,
    )
    print("PlottingMixin plot saved in:", plot_out_dir)
else:
    print("No outputs to visualize yet.")
```
